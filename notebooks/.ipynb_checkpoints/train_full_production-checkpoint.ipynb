# Full Production Run — All Systems, All Experiments

Runs the complete experiment matrix for the manuscript:
- Baselines on all 3 systems
- Benchmark: 7 algorithms × 3 systems × 10 seeds × 3000 episodes  
- Fair comparison: 7 algorithms on 14-bus
- Ablation: 10 variants on 14-bus
- Scalability: stress test on 118-bus

**Estimated time:** 24-48 hours depending on hardware. Each run is cached to disk — safe to interrupt and resume by re-running the cell.


```
%%capture
import subprocess, sys

# Install all dependencies
packages = ["numpy>=1.26", "scipy>=1.11", "torch>=2.1", "pandapower>=3.0",
            "matplotlib>=3.8", "pyyaml>=6.0", "pytest>=7.4"]
for pkg in packages:
    subprocess.check_call([sys.executable, "-m", "pip", "install", pkg, "-q"])

print("All packages installed successfully")

# Verify imports
import numpy as np; import torch; import pandapower
print(f"NumPy {np.__version__}, PyTorch {torch.__version__}, pandapower {pandapower.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

```
import sys, os, time, json
import numpy as np

# Add repo root to path (adjust if notebook is not in notebooks/)
REPO = os.path.abspath(os.path.join(os.getcwd(), ".."))
if REPO not in sys.path:
    sys.path.insert(0, REPO)

# Verify the sim package loads
from sim.config import load
from sim.blackstart_env import BlackStartEnv
from sim.train import SPECS, SCENARIOS, METRICS, run, train, make_env, make_agent, evaluate
from sim.baselines import greedy_policy, random_mask_policy, upper_bound
from sim.stats import summarise

cfg = load(os.path.join(REPO, "configs", "default.yaml"))
print(f"Config loaded: {len(cfg['systems'])} systems")
for s in ("14","39","118"):
    sc = cfg["systems"][s]
    h = sc.get("horizon", cfg["env"]["horizon"])
    dt = sc.get("dt_min", cfg["env"]["dt_min"])
    n_units = len(sc.get("units", [])) or sum(sc.get("unit_counts",{}).values())
    print(f"  IEEE {s}-bus: T={h} steps, dt={dt} min, {n_units} GFM units")
```

```
# Full production run
import subprocess

OUT_FULL = os.path.join(REPO, "sim", "results", "production_10x3000")

# Adjust these based on your hardware
SEEDS = 10
EPISODES = 3000
JOBS = 4  # Number of parallel CPU workers

cmd = [
    sys.executable, "-m", "sim.run_experiments",
    "--seeds", str(SEEDS),
    "--episodes", str(EPISODES),
    "--systems", "14", "39", "118",
    "--jobs", str(JOBS),
    "--out", OUT_FULL,
    "--label", f"production_{SEEDS}x{EPISODES}",
]

print(f"Command: {' '.join(cmd)}")
print(f"\nThis will run {7 * 3 * SEEDS} benchmark + {7 * SEEDS} fair + {10 * SEEDS} ablation training runs")
print(f"Estimated total: {7*3*SEEDS + 7*SEEDS + 10*SEEDS} runs @ ~30-180s each")
print(f"\nStarting...")

result = subprocess.run(cmd, capture_output=False, text=True, cwd=REPO)
print("\nDONE!" if result.returncode == 0 else f"Failed: {result.returncode}")
```

```
# Generate all figures and tables from the production results
import subprocess

fig_dir = os.path.join(OUT_FULL, "figs")
os.makedirs(fig_dir, exist_ok=True)

# Figures
subprocess.run([sys.executable, os.path.join(REPO, "figs", "make_comparison_figures.py"),
                "--results", OUT_FULL, "--out", fig_dir], cwd=REPO)

# Tables
subprocess.run([sys.executable, "-m", "sim.make_tables", "--results", OUT_FULL], cwd=REPO)

print("\nGenerated files:")
for f in sorted(os.listdir(OUT_FULL)):
    print(f"  {f}")
if os.path.exists(fig_dir):
    for f in sorted(os.listdir(fig_dir)):
        print(f"  figs/{f}")
```
